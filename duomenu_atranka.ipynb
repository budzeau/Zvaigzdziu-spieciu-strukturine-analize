```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
import pandas as pd
from matplotlib.ticker import FuncFormatter
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
```

```python
data = pd.read_csv('duomenys.csv', encoding='utf-8') #Hunt katalogas
```

```python
print("Stars:", len(data))
print("Clusters:", data['Name'].nunique())
cluster_counts = data['Name'].value_counts()
print(min(cluster_counts), "-", max(cluster_counts))
```

```python
def comma_formatter(x, pos):
    return f"{x:.1f}".replace(".", ",")
def thousands_formatter(x, pos):
    return f"{int(x/1000)}"
fig, ax = plt.subplots(figsize=(8,5))
ax.hist(data["Prob"], bins=30, color="#8f5fca", edgecolor="black")
ax.axvline(0.9, color="black", linestyle="--", linewidth=2, label = "Prob = 0,9")
ax.legend(loc="upper left")
ax.set_xlabel("Tikimybė")
ax.set_ylabel("Žvaigždžių skaičius, tūkst.")
ax.set_title("Spiečiaus narių tikimybių pasiskirstymas")
ax.xaxis.set_major_formatter(FuncFormatter(comma_formatter))
ax.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
plt.show()
```

```python
members = data[data["Prob"] > 0.9]
def thousands_formatter_xd(x, pos):
    value = x / 1000
    if value == int(value):
        return f"{int(value)}"
    return f"{value:.2f}".replace(".", ",")

fig, ax = plt.subplots(figsize=(8,5))
ax.hist(members["Plx"], bins=20, color="#8f5fca", edgecolor="black")
ax.axvline(0.3, linestyle="--", linewidth=2, color="black", label="Plx = 0,3 mas")
ax.set_xlabel("Paralaksas, mas")
ax.set_ylabel("Žvaigždžių skaičius, tūkst.")
ax.set_title("Paralakso pasiskirstymas")
ax.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
ax.legend(loc="lower right")
axins = inset_axes(ax, width="52%", height="52%", loc="upper right", borderpad=2)
zoom = members[(members["Plx"] >= 20) & (members["Plx"] <= 80)]
axins.hist(zoom["Plx"], bins=20, color="#8f5fca", edgecolor="black")
axins.set_title("Intervale 20–80 mas", fontsize=9)
axins.tick_params(labelsize=8)
axins.yaxis.set_major_formatter(FuncFormatter(thousands_formatter_xd))
plt.show()
```

```python
df = data[(data['Prob'] > 0.9) & (data["Plx"] > 0.3)]
cluster_size = df['Name'].value_counts()
plt.figure()
box = plt.boxplot(cluster_size, vert=True, patch_artist=True)
for patch in box['boxes']:
    patch.set_facecolor("#8f5fca")
    patch.set_edgecolor("black")
for element in ['whiskers', 'caps', 'medians']:
    for line in box[element]:
        line.set_color("black")
plt.title("Spiečių žvaigždžių skaičiaus pasiskirstymas")
plt.ylabel("Žvaigždžių skaičius spiečiuje")
plt.xticks([])
plt.show()
```

```python
df = data[(data['Prob'] > 0.9) & (data["Plx"] > 0.3)]
cluster_size = df['Name'].value_counts()
large_clusters = cluster_size[cluster_size > 20].index
df_filtered = df[df['Name'].isin(large_clusters)]
print("Stars:", len(df_filtered))
print("Clusters:", df_filtered['Name'].nunique())
```
